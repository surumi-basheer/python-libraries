# Abstraction

- Abstraction is a concept in **Object-Oriented Programming ( OOP )** that focuses on **hiding the implementation details** and showing only the necessary featyres of an object.

- It allows  users to focus on **what an object dows** rather than how it does it.

- **Example :**
 - When you use a smartphone you only see the interface ( buttons, icons ) but you don't need to know how the internal circuits work



 - **Why is Abstraction important?**
  - Simplifies complex systems by focusing only on essential aspects.
  - Reduces code complexity and increases reusability.
  - Enhances security by restricting direct acces to implementation details

## **Abstraction in Python Using Abstract Classes**

  - Python provides **abstraction** using **abstract classes** and  **abstract methods** which are defined in the abc module
  - **Abstract class :** A class that contains one or more abstract methods, ti cannot be instantiated directly.
  - **Abstrcat method :** A method that is declared but contains no implementation ( must be implemented by child classses. )

from abc import ABC,abstractmethod

```python
from abc import ABC, abstractmethod
class vehicle(ABC): # abstract class

  @abstractmethod
  def start_engine(self):
    pass # abstract method with no implementation


  @abstractmethod
  def stop_engine(self):
    pass #abstract method with no implementaton

# An abstract class cannot be instantiated directy
v=vehicle() # TypeError: can't instantiate abstract class vehicle with abstract methods start_engine
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
  @abstractmethod
  def calculate_area(self):
    pass


class circle(Shape):

  def __init__(self,radius):
    self.radius = radius

  def calculate_area(self):
    print(f"circle area is : {3.14 * (self.radius)**2 }")


class Rectangle(Shape):

  def __init__(self, width, height):
    self.width = width
    self.height = height

  def calculate_area(self):
    print(f"Rectangle area is : {self.width * self.height}")

cir = circle(10)
rect= Rectangle(5,10)

cir.calculate_area()
print()
rect.calculate_area()
```

```python
import threading
import time

# Function for the chef's work(Thread 1)
def chef_work():
  for i in range(5): # chef prepares 5 dishes
      print(f"chef: preparing dish {i + 1}...")
      time.sleep(2) # simulate time taken to prepare a dish
      if i == 2: # Assistant starts washing utensils during the 3rd dish
          print("chef: calling assistant to wash utensils.")
          time.sleep(1) # simulate time taken to call the assistant
          assistant_thread = threading.Thread(target=assistant_work)
          assistant_thread.start()
          assistant_thread.join()  #wait for the asistant to finish
          print("chef assistant done. resuming dish prepration.")
  print("chef: All dishes are prepared.")

# function for the assistant's work(Thread 2)
def assistant_work():
  for j in range(3): # assistant washes 3 utensils
      print(f" Assistant: washing utensil {j +1}...")
      time.sleep(2) # simulate time taken to wash each utensil
  print(" Assistant: All utensils are clean.")

print("kitchen: starting work.")
chef_thread = threading.Thread(target=chef_work)
chef_thread.start()
chef_thread.join() # wait for the chef to complete their work
print("kitchen: All work completed.")
```

### Qustion
Write a Python program using threading to simulate music playback and lyrics downloading:

Music playback runs in one thread for 5 seconds, printing a message each second.
During the 3rd second, a separate thread starts to simulate downloading lyrics, which takes 3 seconds.
Ensure the music playback waits for the lyrics to finish downloading before continuing.

```python
import threading
import time

def music_playback():
  print("Music player: Starting music playback")
  for x in range(5):
    print(f"Music player: playing music... {x+1} seconds")
    time.sleep(2)

    if x ==2:
      print("Music player:Downloading lyrics in the background.")
      time.sleep(3)
      lyrics_thread = threading.Thread(target=lyrics_downloading)
      lyrics_thread.start()
      lyrics_thread.join()
      print("Music player: Lyrics downloaded")

def lyrics_downloading():
  for i in range (3):
    print(f"Downloading lyrics... {i+1} seconds")
    time.sleep(1)
  print("Lyrics download completed.")

musicplayer_thread = threading.Thread(target=music_playback)
musicplayer_thread.start()
musicplayer_thread.join()
print("Music player: playback ended")
```

## RegEx in Python

```python
# search the string to see if it starts with "The" and ends with "Spain":

import re
txt = "The rain in spain"
x = re.search("^The.*spain$",txt)
if x:
  print("YES! We have a match!")
else:
  print("No match")
```

## The findall() Function

```python
# The findall() function returns a list containing all matches.
#print a list of all matches:
import re
txt = "The rain in spain"
x = re.findall("ai", txt)
print(x)
```

```python
# Return an empty list no match was found:
import re
txt = "The rain in spain"
# check if "Portugal" is in the string:
x = re.findall("portugal",txt)
print(x)
if (x):
  print("Yes,there is at least one match!")
else:
  print("No match")
```

## The search()Function

- If there is more than one match,only the first occurence of the match will be returned:

```python
# Search for the first white - space character in the string:

import re
txt = "The rain in spain"
x = re.search("\s",txt)
print("The first white-space character is located in position:", x.start())
```
