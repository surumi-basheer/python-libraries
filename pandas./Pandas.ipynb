## Series
- A Pandas Series is like a column in a table.
- It is a one-dimensional array holding data of any type.

```python
# Create a simple panda series from a list:
import pandas as pd
a = [1,7,2]
myvar = pd.Series(a)
print(myvar)
```

## Labels
(index number)

```python
print(myvar[0])
```

## Create Labels
- With the index arguments,you can name your own labels.

```python
# Crate your own labels:
import pandas as pd
a =[1,7,2]
myvar = pd.Series(a,index =["x","y","z"])
print(myvar)
```

```python
# Return the value of"y":
print(myvar["y"])
```

# Merge

```python
tcs={'id':[101,102,103,106],
      'name':['Mithun','Dipin','jose','maneesh'],
      'age':[25,26,27,29]}
wipro={'id':[101,102,103,105],
      'name':['dev','sreenath','sreerag','varun'],
      'salary':[25000,34000,37000,20000]}
tcs_emp=pd.DataFrame(tcs)
wipro_emp=pd.DataFrame(wipro)
```

```python
pd.merge(tcs_emp,wipro_emp,on='id')
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',suffixes=['_left','_right'])
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='outer')
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='outer',indicator=True)
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='left')
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',how="right")
```

## Cross tab or contigency table
- In pandas,pd.crosstab() is used to compute a cross-tabulation of two or more factors. It’s often used for analyzing categorical variables, similar to a contingency table in statistics.

syntax:

pd.crosstab(index, columns, values=None, aggfunc=None, margins=False, margins_name='All')



```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/dataset/survey_crosstab (1).csv")
df
```

```python
# Create cross-tabulation
ct = pd.crosstab(df['Gender'], df['Preferred_Device'])

print(ct)
```

```python
# Cross-tab of AgeGroup vs Preferred_Device with totals:

ct2 = pd.crosstab(df['AgeGroup'], df['Preferred_Device'], margins=True, margins_name='Total')
print(ct2)
```
